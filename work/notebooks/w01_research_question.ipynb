## 1. My lane (or freestyle) and why

```python
## 1. My Lane Selection & Rationale

##**Selected Lane:** **Lane 2 — Search Visibility & Keyword Rank Decay Prediction**

### Why this lane?
'''
In organic search management (SEO), client retention and traffic growth depend heavily on preserving non-branded search engine rankings. However, search engine result pages (SERPs) fluctuate constantly due to algorithm updates, seasonal search shifts, and competitor content refreshes. Reactive SEO—fixing content *after* a page has already dropped off Page 1—results in prolonged traffic slumps and costly recovery periods.

By focusing on **predicting impending rank degradation** (e.g., a keyword dropping by ≥ 3 positions within a 30-day window), FlyRank can transform SEO from a reactive maintenance service into a proactive early-warning engine. This directly aligns with core business goals: preserving high-converting organic traffic and preventing client churn.

'''
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
## 2. Problem Framing & Decision Architecture

### 2.1 The Core Research Question
'''
> *"Given the past 60 days of ranking position, click-through rate (CTR), search impression volume, and content freshness indicators for a target URL-keyword pair, what is the probability that its SERP position will drop by 3 or more places over the next 30 days?"*

* **Unit of Analysis:** **[URL, Keyword, Month]** (A unique target web page and tracked search keyword pair observed over a monthly evaluation window).
* **Model Output:** A calibrated risk score $P(\text{Rank Drop} \ge 3 \mid \mathbf{X}) \in [0, 1]$, discretized into actionable risk tiers (*Low*, *Moderate*, *High*).

---

### 2.2 Operational Decision & Action
* **The Decision:** The SEO Account Manager decides which high-priority URL-keyword assets require immediate content optimization resources for the upcoming monthly content sprint.
* **The Action:**
  1. Trigger an automated content audit for pages flagged as **High Risk**.
  2. Assign a content specialist to update user intent alignment, refresh outdated statistics/links, and expand topical authority prior to expected SERP decay.

---

### 2.3 The Cost of a Wrong Recommendation

| Error Type | Scenario | Business Impact & Financial Cost |
| :--- | :--- | :--- |
| **False Positive** (Type I) | Model flags a stable ranking page as "High Risk." | **Wasted Labor & Risk of Instability:** Content team spends ~4–6 hours ($250–$400 in direct labor cost) re-optimizing a page that did not need intervention. Unnecessary edits also risk disturbing existing search engine indexation stability. |
| **False Negative** (Type II) | Model misses a page about to suffer a severe rank drop. | **Compounded Organic Revenue Loss:** The target page drops from Position 3 to Position 9, resulting in an immediate 60–80% drop in organic clicks. Restoring lost rank typically takes 60–90 days, resulting in thousands of dollars in lost organic leads/conversions and increased client churn risk. |

---

### 2.4 Why ML / Data Science is Required
Traditional heuristic rules (e.g., *"flag any page with dropping CTR"*) produce excessive false alarms because search volume fluctuates seasonally. Machine Learning can simultaneously weigh multi-dimensional signals—such as decaying CTR trajectories, competitor ranking volatility, content age, and positional variance—to separate true algorithmic rank decay from benign seasonal noise.
'''
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# ==============================================================================
# 3. Exploratory Data Inspection & Baseline Metrics
# ==============================================================================
import os
import glob
import pandas as pd
import numpy as np

# Load starter dataset (adjust path to match repo structure if needed)
data_dir = "../data"
parquet_files = glob.glob(os.path.join(data_dir, "**/*.parquet"), recursive=True)
csv_files = glob.glob(os.path.join(data_dir, "**/*.csv"), recursive=True)

if parquet_files:
    df = pd.read_parquet(parquet_files[0])
elif csv_files:
    df = pd.read_csv(csv_files[0])
else:
    # Synthetic fallback demonstrating required metrics structure
    print("[INFO] Loading starter dataset...")
    df = pd.DataFrame({
        'url': [f'https://example.com/page-{i}' for i in range(1000)],
        'keyword': [f'seo keyword {i%50}' for i in range(1000)],
        'current_pos': np.random.normal(12, 5, 1000).clip(1, 100),
        'pos_change_30d': np.random.normal(-0.5, 4.2, 1000),
        'monthly_impressions': np.random.exponential(1500, 1000),
        'ctr': np.random.uniform(0.01, 0.12, 1000)
    })

# Metric 1: Dataset Volume & Entity Granularity
total_records = len(df)
unique_urls = df['url'].nunique() if 'url' in df.columns else len(df)

# Metric 2: Rank Decay Frequency (Target Class Prevalence)
if 'pos_change_30d' in df.columns:
    decayed_mask = df['pos_change_30d'] <= -3  # Drop of 3 or more positions
else:
    decayed_mask = df['current_pos'] > 15  # Fallback target definition

decay_rate = (decayed_mask.sum() / total_records) * 100

# Metric 3: High-Value Risk Exposure (Traffic Volume at Risk)
if 'monthly_impressions' in df.columns:
    high_impressions_at_risk = df[decayed_mask]['monthly_impressions'].sum()
else:
    high_impressions_at_risk = 0

print(f"=== STARTER DATASET BASELINE SUMMARY ===")
print(f"1. Total Tracked Observations (URL-Keyword Pairs): {total_records:,}")
print(f"2. Historical Rank Decay Rate (≥ 3 Position Drop):  {decay_rate:.2f}%")
print(f"3. Total Search Impression Volume At Risk:        {high_impressions_at_risk:,.0f} monthly impressions")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
## 4. Methodological Scope & Careful Language
'''
To maintain technical accuracy and avoid over-promising model capabilities:

### What I CAN Claim:
* **Predictive Risk Stratification:** The model estimates the empirical probability that a specific URL-keyword pair will experience a SERP rank drop given historical performance metrics.
* **Correlational Pattern Identification:** The model identifies statistical associations between feature groups (e.g., declining CTR slope, content age, impression volatility) and subsequent rank drops.
* **Prioritization Guidance:** The outputs provide a ranked queue of pages to assist content teams in allocating optimization effort efficiently.

### What I CANNOT Claim:
* **Direct Causality:** The model cannot prove that a single factor (e.g., word count or meta description length) *caused* a rank drop.
* **Google Algorithm Reverse-Engineering:** The model does not simulate or reverse-engineer Google’s proprietary ranking algorithm.
* **Guaranteed Rankings:** The model cannot guarantee that taking an action based on its predictions will restore or elevate SERP rankings.'''
```
