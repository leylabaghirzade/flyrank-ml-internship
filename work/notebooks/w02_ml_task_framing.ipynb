## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## 1. My Lane as an ML Task
'''
**Selected Lane:** **Lane 2 — Search Visibility & Keyword Rank Decay Prediction**

### ML Task Formulation
This problem is framed as a **Binary Classification & Risk Scoring Task**.
* **Model Objective:** For a given target page-keyword pair over an evaluation window, estimate the calibrated probability $P(Y = 1 \mid \mathbf{X})$ that the keyword will experience significant rank decay ($\ge 3$ SERP position drop) in the subsequent 30 days.
* **Output Type:** A continuous probability score $\in [0, 1]$ mapped into actionable priority tiers (*Low Risk*, *Medium Risk*, *High Risk*).'''
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
'''## 2. Target Definition & Proxy Strategy

### Target Variable (`is_declining_label`)
* **Binary Label Formulation ($Y$):**
  $$Y = \begin{cases} 1, & \text{if } (\text{Position}_{t+30} - \text{Position}_t) \ge +3 \text{ (Rank dropped by } \ge 3 \text{ spots)} \\ 0, & \text{otherwise} \end{cases}$$
  *(Note: Higher position number in Google SERP indicates a lower rank, e.g., Position 1 is top spot, Position 10 is bottom of Page 1).*

### Proxy & Data Leakage Protections
* **Ground Truth Proxy:** Since true algorithmic penalty signals are unobservable, future 30-day position movement ($\Delta \text{Position}_{t \to t+30}$) serves as our target proxy.
* **Strict Feature Leakage Guard:** Neither `trend_direction` nor `trend_pct` may EVER be used as input features, as they directly derive from the target label calculation. Features must strictly leverage historical observations available up to time $t$.'''
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
'''## 3. Evaluation & Success Metrics

### Primary Metric: **Precision@K (e.g., Precision@50)**
* **Why Precision@K?** An SEO operations team has limited capacity—they can only perform roughly 50 deep content audits per monthly sprint. High precision in the top $K$ flagged items guarantees that team efforts are concentrated on pages genuinely experiencing rank decay rather than wasting labor on false alarms.

### Secondary Metrics:
1. **PR-AUC (Area Under Precision-Recall Curve):** Evaluates rank ordering performance across all risk thresholds under class imbalance.
2. **Recall @ 80% Precision Threshold:** Measures what percentage of true rank decays we catch while guaranteeing an acceptable false-positive budget.

| Metric | Target Baseline | Operational Meaning |
| :--- | :--- | :--- |
| **Precision@50** | $\ge 0.70$ (~3x lift over baseline) | At least 35 of the 50 flagged pages actively require content updates. |
| **PR-AUC** | $\ge 0.65$ | Robust ranking capability across fluctuating monthly volumes. |'''
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
'''## 5. Why ML Beats Fixed Heuristics

Traditional SEO workflows rely on hardcoded static rules (e.g., *"Flag any page where CTR drops by 15% month-over-month"*). These rules fail consistently in real-world organic search environments for three reasons:

1. **Seasonal & Volatility Noise:** A sudden 20% drop in search impressions for a seasonal query (e.g., "tax filing software" in June) is normal market behavior, not organic decay. A static rule triggers expensive false alarms during off-peak seasons.
2. **Non-Linear Multi-Variable Interactions:** Rank decay is rarely caused by a single metric. A page with a slight CTR drop might remain stable if its impression volume is increasing, while a page with stable CTR might decay if competitor volatility rises. ML tree ensembles (e.g., Random Forest, XGBoost) capture these non-linear interactions automatically.
3. **Capacity-Aware Priority Scoring:** Fixed rules return an unranked binary list (either a page breaches the threshold or it doesn't). ML outputs a calibrated probability score, allowing the operations team to dynamically adjust the top-$K$ threshold to fit available content team capacity.'''
```
